```python
import os
import asyncio
import time

from dotenv import load_dotenv
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain.agents import create_agent

load_dotenv()
```

```python
client = MultiServerMCPClient(
        {
            "Math-mcp": {
                "transport": "stdio",  # Local subprocess communication
                "command": "python",
                # Absolute path to your math_server.py file
                "args": ["/home/bipin/Documents/genai-july/agentic ai-class/mcp_servers/math_mcp.py"],
            },
            "weather": {
                "transport": "http",  # HTTP-based remote server
                # Ensure you start your weather server on port 8000
                "url": "http://127.0.0.1:8000/mcp",
            }
        }
    )
```

```python
import sys
import builtins

sys.stdout = builtins.open(1, 'w', buffering=1, closefd=False)
sys.stderr = builtins.open(2, 'w', buffering=1, closefd=False)
```

```python
tools = await client.get_tools()
```

```python
tools
```

```python
agent = create_agent(
        "openai:gpt-5.4-nano",
        tools  
    )
math_response = await agent.ainvoke(
    {"messages": [{"role": "user", "content": "what's (3 + 5) x 12?"}]}
)
```

```python
for message in math_response["messages"]:
    message.pretty_print()
```

```python
weather_response = await agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in nyc?"}]}
)
print(math_response)
print(weather_response)
```
