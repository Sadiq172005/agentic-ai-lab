```python
import os
import asyncio
import time

from dotenv import load_dotenv
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain.agents import create_agent

load_dotenv()
```

```python
client = MultiServerMCPClient(
        {
            
            "Patient": {
                "transport": "http",  # HTTP-based remote server
                # Ensure you start your weather server on port 8000
                "url": "http://127.0.0.1:8003/mcp",
            }
        }
    )
```

```python
tools = await client.get_tools()
```

```python
tools
```

```python
agent = create_agent(
        "openai:gpt-5.4-nano",
        tools  
    )
patientagent_response = await agent.ainvoke(
    {"messages": [{"role": "user", "content": "give the patient ids whose blood group are o+ve and age is greater than 30"}]}
)
```

```python
print(patientagent_response["messages"][-1].content)
```

```python
for message in patientagent_response["messages"]:
    message.pretty_print()
```
