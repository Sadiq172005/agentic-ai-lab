# MCP Agent - All Servers

Before running this notebook, start each `http` transport server in its own terminal (from the project root):

```
python mcp_servers/setup_patient_db.py   # run once, creates patients.db
python mcp_servers/weather_mcp.py        # port 8000
python mcp_servers/pharmacist_mcp.py     # port 8001
python mcp_servers/restaurant_mcp.py     # port 8002
python mcp_servers/patient_mcp.py        # port 8003
```

`math_mcp.py` uses `stdio` transport, so the client launches it automatically - no need to start it manually.

````python
import os

from dotenv import load_dotenv
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain.agents import create_agent

load_dotenv()
````

````python
client = MultiServerMCPClient(
    {
        
        "weather": {
            "transport": "http",  # HTTP-based remote server
            # Ensure you start your weather server on port 8000
            "url": "http://127.0.0.1:8000/mcp",
        },
        "pharmacist": {
            "transport": "http",
            # Ensure you start your pharmacist server on port 8001
            "url": "http://127.0.0.1:8001/mcp",
        },
        "restaurant": {
            "transport": "http",
            # Ensure you start your restaurant server on port 8002
            "url": "http://127.0.0.1:8002/mcp",
        },
        "patient": {
            "transport": "http",
            # Ensure you start your patient server on port 8003
            "url": "http://127.0.0.1:8003/mcp",
        },
    }
)
````

````python
tools = await client.get_tools()
````

````python
tools
````

````python
agent = create_agent(
    "openai:gpt-5.4-nano",
    tools
)
````

## Math server

````python
math_response = await agent.ainvoke(
    {"messages": [{"role": "user", "content": "what's (3 + 5) x 12?"}]}
)
for message in math_response["messages"]:
    message.pretty_print()
````
