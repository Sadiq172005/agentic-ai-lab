```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
load_dotenv()
```

```python
from pprint import pprint
```

```python
@tool
def add_num(a: int, b: int) -> int:
    """this function adds two numbers"""
    y = a + b
    return y
@tool
def multiply_nums(a: int, b: int) -> int:
    """this function multiplies two numbers"""
    y = a * b
    return y
@tool
def subtract_nums(a: int, b: int) -> int:
    """this function subtracts two numbers"""
    y = a - b
    return y
@tool
def divide_nums(a: int, b: int) -> float:
    """this function divides two numbers"""
    y = a / b
    return y

@tool
def search(query: str) -> str:
    """Searches the web for a given query and returns the results."""
    # Here you would implement the actual search logic, e.g., using an API.
    return f"Results for: {query}"
```

```python
tools = [add_num, multiply_nums, subtract_nums, divide_nums, search]
```

```python
agent = create_agent(model="openai:gpt-5.4-nano", tools=tools)
```

```python
result = agent.invoke({"messages": [{"role": "user", "content": "what is the sum of 56*45-89*56 +88 ?"}]})
```

```python
import requests
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get the current weather for a city."""
    # Step 1: convert city name to lat/lon
    geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city}"
    geo_res = requests.get(geo_url).json()

    if "results" not in geo_res:
        return f"Could not find location: {city}"

    lat = geo_res["results"][0]["latitude"]
    lon = geo_res["results"][0]["longitude"]

    # Step 2: fetch weather for that lat/lon
    weather_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current_weather=true"
    weather_res = requests.get(weather_url).json()

    temp = weather_res["current_weather"]["temperature"]
    wind = weather_res["current_weather"]["windspeed"]

    return f"The weather in {city} is {temp}°C with wind speed {wind} km/h."

result = get_weather.invoke({"city": "Kashmir"})
print(result)
```

```python
from langchain.agents import create_agent
from langchain.tools import tool
from pprint import pprint

@tool
def search(query: str) -> str:
    """Search for information."""
    return f"Results for: {query}"

agent = create_agent(
    model="openai:gpt-5.4-nano",
    tools=[search],
    system_prompt="You are a helpful research assistant."
)

result = agent.invoke({
    "messages": [
        {"role": "user", "content": "Research AI safety trends"}
    ]
})

for message in result['messages']:
    message.pretty_print()
```
