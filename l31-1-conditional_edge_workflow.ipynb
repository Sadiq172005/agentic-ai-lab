```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END


# -------------------------
# 1. Define the State
# -------------------------

class RestaurantState(TypedDict):
    people: int
    time: str
    available: bool
    message: str
```

```python
def check_availability(state: RestaurantState):
    print("Checking table availability...")

    # Simple hard-coded example
    if state["people"] <= 4 and state["time"] == "8 PM":
        available = True
    else:
        available = False

    return {
        "available": available
    }
```

```python
def book(state: RestaurantState):
    print("Booking table...")

    return {
        "message": f"Table booked for {state['people']} people at {state['time']}."
    }
```

```python

def unavailable(state: RestaurantState):
    print("Table is not available.")

    return {
        "message": f"Sorry, no table available for {state['people']} people at {state['time']}."
    }
```

```python
def check_result(state: RestaurantState):

    if state["available"]:
        return "do-booking"

    return "unavailable-now"
```

```python
graph = StateGraph(RestaurantState)
graph.add_node("check-table", check_availability)
graph.add_node("table-booking", book)
graph.add_node("table-unavailable", unavailable)

graph.add_edge(START, "check-table")
graph.add_edge( "table-booking", END)
graph.add_edge("table-unavailable", END)

graph.add_conditional_edges("check-table",
                           check_result,
                           { "do-booking": "table-booking", "unavailable-now": "table-unavailable" })
```

```python
app = graph.compile()
from IPython.display import Image, display

display(
    Image(
        app.get_graph().draw_mermaid_png()
    )
)
```

```python
result = app.invoke({
    "people": 5,
    "time": "8 PM",
})


print(result)
```
