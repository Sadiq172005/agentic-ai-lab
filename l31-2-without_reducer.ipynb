```python
from langgraph.graph import StateGraph, START, END
from typing_extensions import TypedDict

class State(TypedDict):
    results: list[str]

def customer_node(state):
    return {
        "results": ["Customer: Bipin"]
    }


def account_node(state):
    return {
        "results": ["Account: Savings"]
    }

def final_node(state: State):
    print("Final state:", state)
    return {}
```

```python
builder = StateGraph(State)

builder.add_node("customer", customer_node)
builder.add_node("account", account_node)
builder.add_node("final", final_node)
```

```python
# START runs both nodes in parallel
builder.add_edge(START, "customer")
builder.add_edge(START, "account")

# Both go to final
builder.add_edge("customer", "final")
builder.add_edge("account", "final")

builder.add_edge("final", END)
graph = builder.compile()
```

```python
from IPython.display import Image, display

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)
```

```python
result = graph.invoke({
    "results": ["hi"]
})

print(result)
```
